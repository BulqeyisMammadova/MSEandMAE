```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error,mean_absolute_percentage_error
```

```python
df = pd.read_csv("insurance.csv")
```

```python
df.head(3)
```

```python
cols_to_encode = ["sex", "smoker", "region"]
if all(col in df.columns for col in cols_to_encode):
    df = pd.get_dummies(df, columns=cols_to_encode, drop_first=True)
```

```python
df.head(3)
```

```python
y = df["charges"]
X = df.drop("charges", axis=1)
```

```python
lm = LinearRegression()
model = lm.fit(X,y)
```

```python
model.score(X,y)
```

```python
df_error = pd.DataFrame()
```

```python
df_error['y']=y
```

```python
df_error
```

```python
y_estimate = model.predict(X)
```

```python
df_error['estimate']=y_estimate
```

```python
df_error
```

```python
df_error['error']=df_error['y']-df_error['estimate']
```

```python
df_error
```

```python
df_error['MSE'] = df_error['error']**2
```

```python
df_error
```

```python
df_error["MAE"] = abs(df_error['error'])
```

```python
df_error['Percent_error'] = np.abs((y-y_estimate)/y)
```

```python
df_error
```

```python
df_error.mean()
```

```python
mean_squared_error(y,y_estimate)
```

```python
mean_absolute_error(y,y_estimate)
```

```python
mean_absolute_percentage_error(y,y_estimate)
```
